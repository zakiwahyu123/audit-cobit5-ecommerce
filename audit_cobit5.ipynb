```python

# ==========================================
# SEL 3: PERHITUNGAN HASIL AUDIT COBIT 5
# ==========================================

def hitung_hasil_audit(df):
    # Mengambil pertanyaan yang sudah diberi skor
    df_dinilai = df[df["Skor"].notna()].copy()

    total_pertanyaan = len(df)
    jumlah_dinilai = len(df_dinilai)
    belum_dinilai = total_pertanyaan - jumlah_dinilai

    if jumlah_dinilai > 0:
        rata_rata = df_dinilai["Skor"].mean()
        persentase = (rata_rata / 3) * 100
    else:
        rata_rata = 0
        persentase = 0

    print("========== HASIL AUDIT COBIT 5 ==========")
    print("Total pertanyaan :", total_pertanyaan)
    print("Sudah dinilai    :", jumlah_dinilai)
    print("Belum dinilai    :", belum_dinilai)
    print("Rata-rata skor   :", round(rata_rata, 2))
    print("Persentase       :", round(persentase, 2), "%")

    # Rekapitulasi setiap prinsip
    print("\n========== REKAP PER PRINSIP ==========")

    rekap = df_dinilai.groupby("Prinsip").agg(
        Jumlah_Dinilai=("Skor", "count"),
        Rata_Rata=("Skor", "mean")
    ).reset_index()

    if not rekap.empty:
        rekap["Persentase"] = (
            rekap["Rata_Rata"] / 3 * 100
        ).round(2)

        display(rekap)
    else:
        print("Belum ada data penilaian.")

    return rekap


# Menjalankan perhitungan
rekap_audit = hitung_hasil_audit(df_audit)
```

```python

# ==========================================
# SEL 4: VISUALISASI GRAFIK HASIL AUDIT
# ==========================================

# Menghitung rata-rata skor setiap prinsip
rata_prinsip = df_audit.groupby("Prinsip")["Skor"].mean()

# Mengisi prinsip yang belum memiliki penilaian dengan 0
rata_prinsip = rata_prinsip.reindex(
    df_audit["Prinsip"].unique(),
    fill_value=0
)

# Membuat grafik batang
plt.figure(figsize=(12, 6))

rata_prinsip.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Grafik Hasil Audit COBIT 5")
plt.xlabel("Prinsip COBIT 5")
plt.ylabel("Rata-rata Skor")
plt.ylim(0, 3)

plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

print("Visualisasi grafik berhasil dibuat!")
```

```python

# ==========================================
# SEL 5: REKAPITULASI HASIL AUDIT
# ==========================================

print("==========================================")
print("REKAPITULASI HASIL AUDIT COBIT 5")
print("SISTEM INFORMASI E-COMMERCE")
print("==========================================")

# Menampilkan seluruh data audit
display(df_audit)

# Rekap jumlah pertanyaan setiap prinsip
rekap_jumlah = df_audit.groupby("Prinsip").agg(
    Total_Pertanyaan=("No", "count"),
    Sudah_Dinilai=("Skor", "count")
).reset_index()

rekap_jumlah["Belum_Dinilai"] = (
    rekap_jumlah["Total_Pertanyaan"] -
    rekap_jumlah["Sudah_Dinilai"]
)

print("\nREKAP BERDASARKAN LIMA PRINSIP COBIT 5")
display(rekap_jumlah)

print("\nRekapitulasi berhasil dibuat!")
```
